```python
import pandas as pd
import sys
import os
import glob

# Write your code below.
sys.path.append(os.getenv('PRICE_DATA'))

from utils.logger import get_logger
_logs = get_logger(__name__)

parquet_files = glob.glob(os.path.join(os.getenv("PRICE_DATA"), "**","*.parquet"), recursive=True)

```

```python
parquet_files
```

```python
dfs = []

for path in parquet_files:
  df = dd.read_parquet(path, columns=['Close', 'Adj Close','High','Low'])
  ticker = os.path.basename(os.path.dirname(path)).split('_')[0]

# Add lagged columns (1-period lag)
  df['Close_lag_1'] = df['Close'].shift(1)
  df['Adj_Close_lag1'] = df['Adj Close'].shift(1)
  df['returns'] = (df['Close']/ df['Close_lag_1'])-1
  df['hi_lo_range']=df['High']-df['Low']
  df['ticker'] = ticker
  dfs.append(df)

dd_feat=dd.concat(dfs, axis=0)
# Preview the result
print(dd_feat.head())

```

For each ticker and using Dask, do the following:

+ Add lags for variables Close and Adj_Close.
+ Add returns based on Close:
    
    - `returns`: (Close / Close_lag_1) - 1

+ Add the following range: 

    - `hi_lo_range`: this is the day's High minus Low.

+ Assign the result to `dd_feat`.

(4 pt)

```python
# Write your code below.
df=dd_feat.compute()
df['Return_Moving_avg_10days'] = df['returns'].rolling(window=10).mean()
df
```
